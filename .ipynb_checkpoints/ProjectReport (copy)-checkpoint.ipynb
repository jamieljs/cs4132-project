## Motivation & Background<a id='motivbg'/>
There is increasing emphasis on the environmental issue of food waste locally and worldwide. Through investigating food production, trade and waste trends, and how these relate to factors such as GDP per capita and population size, I hope to learn more about food production and consumption worldwide, as well as gain insight into some possible reasons for food waste that need to be tackled.

For instance, increased urbanisation decreases the amount of space allocated for food production, but new technologies and methods developed such as vertical farming could boost food production, hence I would like to analyse trends in global and local food production (e.g. agriculture or aquaculture).

Moreover, there could be many factors affecting how much food a country needs and how much it imports or exports (and whether this is too much or too little), such as the country's demographics or economic status. I hope analysing such patterns will allow me to learn more about food consumption patterns and reasons behind food waste.

### Data Cleaning<a id='cleaning'/>

For the WorldBank data, there are rows containing information from groups of regions/countries. We remove these rows.

```python
def average_ranges(row):
    if row.loss.find('+/-') != -1: # middle +- uncertainty
        row.loss = row.loss.split('+')[0]
    elif row.loss.find('-') != -1: # range
        loss_range = row.loss.split('-')
        row.loss = str((float(loss_range[0]) + float(loss_range[1]))/2)
    return row

df_waste = df_waste.apply(average_ranges,axis='columns')
```

```python
def replace_characters(row):
    if row.loss.find('â€“') != -1:
        loss_split = row.loss.split('â€“')
        row.loss = loss_split[0] + '.' + loss_split[1]
    return row

df_waste = df_waste.apply(replace_characters,axis='columns')
```

###### A. Over the years?
We draw point plots of the area used for agriculture (which is likely to be related to amount of agriculture), percentage of country's area used for agriculture and aquaculture production, against time.

We also draw side-by-side boxplots for the percentage area used for agriculture over the years.

```python
def eda_pointplots(df, countries, ylabel, title, figsize):
    mask = df['Country Name']==countries[0]
    for i in range(1,len(countries)):
        mask = mask|(df['Country Name']==countries[i])
    # each column is a different country, each row is a different year; drop the country code
    df_plot = df[mask].T.drop(['Country Code'])
    # set country names as column labels
    df_plot.columns = df_plot.loc['Country Name']
    df_plot = df_plot.drop(['Country Name'])
    # make years a column and not the indices
    df_plot = df_plot.reset_index()
    # unpivot the df
    df_plot = df_plot.melt(id_vars='index',var_name='country',value_name='values')
    plt.figure(figsize=figsize)
    sns.pointplot(x='index',y='values',hue='country',data=df_plot)
    plt.title(title)
    plt.ylabel(ylabel)
    plt.xlabel('Years')
    plt.xticks(ticks=np.arange(0,61,5))

def eda_boxplots(df):
    # remove world data to avoid skewing the axis scale
    # drop country name and country code columns
    countries_only = df[df['Country Name']!='World'].drop(['Country Name','Country Code'],axis=1)
    countries_only.plot(kind='box',figsize=(15,8))
    plt.xticks(ticks=np.arange(0,61,5))
```

###### B. Across various countries?
We plot choropleth maps to visualise how the geography of various countries might affect the amount of agricultural land and aquaculture production. (e.g. land-locked countries vs countries near the coast, or differences across continents)

```python
def eda_q1b_draw_map(df, legend):
    world_map = folium.Map(location=[0,0], zoom_start=1)
    folium.Choropleth(
        geo_data=world_geo,
        data=df,
        columns=['Country Name', 'Value'],
        key_on='feature.properties.name',
        fill_color='YlGn', 
        fill_opacity=0.7, 
        line_opacity=0.2,
        legend_name=legend
    ).add_to(world_map)
    return world_map
```

We also draw scatterplots for the worldwide GDP per capita against agricultural area and aquaculture production over time.

```python
def eda_q1cd_scatter(ax, x, y, start_year, end_year, xlabel, ylabel):
    title = 'Worldwide ' + ylabel + ' vs ' + xlabel + ' from ' + start_year + ' to ' + end_year
    x = x.loc[x['Country Name']=='World', start_year:end_year].T
    y = y.loc[y['Country Name']=='World', start_year:end_year].T
    mask = (x.notna().all(axis=1))&(y.notna().all(axis=1))
    x = x[mask].values.flatten()
    y = y[mask].values.flatten()
    print(title + ": " + str(stats.pearsonr(x,y)))
    ax.scatter(x,y,alpha=0.5)
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.set_title(title)
```

```python
def eda_q1c_scatter():
    fig, axs = plt.subplots(4,1,figsize=(15,24))
    eda_q1cd_scatter(axs[0], df_gdp, df_agri, '1961', '2020', 'GDP per capita', 'agricultural area')
    eda_q1cd_scatter(axs[1], df_gdp, df_agri, '1961', '1991', 'GDP per capita', 'agricultural area')
    eda_q1cd_scatter(axs[2], df_gdp, df_agri, '1992', '2020', 'GDP per capita', 'agricultural area')
    eda_q1cd_scatter(axs[3], df_gdp, df_agri, '1961', '2020', 'GDP per capita', 'aquaculture production')

eda_q1c_scatter()
```

###### How are the amounts of agriculture and aquaculture related?

For worldwide data as well as various countries, we draw a scatter plot of aquaculture production against the area used for agriculture, where each point is a different year.

```python
# each point is a different year
def eda_q2_scatter_time(country, start_year, end_year, title):
    country_agri = df_agri.loc[(df_agri['Country Name']==country),start_year:end_year].T.copy()
    country_aqua = df_aqua.loc[(df_aqua['Country Name']==country),start_year:end_year].T.copy()
    df_country = pd.concat([country_agri,country_aqua],axis=1)
    df_country.columns=['Agriculture','Aquaculture']
    df_country.dropna(axis=0, how='any', inplace=True)
    print(title + ": " + str(stats.pearsonr(df_country['Agriculture'],df_country['Aquaculture'])))
    plt.figure(figsize=(12,6))
    plt.scatter(df_country['Agriculture'],df_country['Aquaculture'],alpha=0.5)
    plt.xlabel('Agricultural area (sq. km)')
    plt.ylabel('Aquaculture production (metric tons)')
    plt.title(title)
    plt.show()
```

```python
# each point is a different country
def eda_q2_scatter_country(year, title, xlim, ylim):
    year_agri = df_agri.loc[(df_agri['Country Name']!='World'),year].copy()
    year_aqua = df_aqua.loc[(df_aqua['Country Name']!='World'),year].copy()
    df_year = pd.concat([year_agri,year_aqua],axis=1)
    df_year.columns=['Agriculture','Aquaculture']
    df_year.dropna(axis=0, how='any', inplace=True)
    df_year = df_year.loc[(df_year['Agriculture']<xlim)&(df_year['Aquaculture']<ylim)]
    print(title + ": " + str(stats.pearsonr(df_year['Agriculture'],df_year['Aquaculture'])))
    plt.figure(figsize=(12,6))
    plt.scatter(df_year['Agriculture'],df_year['Aquaculture'],alpha=0.5)
    plt.xlabel('Agricultural area (sq. km)')
    plt.ylabel('Aquaculture production (metric tons)')
    plt.title(title)
    plt.show()
```

- agricultural area, percentage area for agriculture and aquaculture production
- food imports, food exports and net food trade

- population size

```python
def eda_q4c_scatter():
    y_df = df_waste.groupby(['country','year','crop_group'])[['loss']].median().reset_index()
    xlabels = ['agricultural area','percentage area for agriculture','aquaculture production',
              'food imports','food exports','net food trade','GDP per capita','population size']
    x_dfs = [df_agri, df_agri_pc, df_aqua, df_food_in, df_food_out, net_trade, df_gdp, df_pop]

    fig, axs = plt.subplots(8, 1, figsize=(12,54))

    for i in range(8):
        x_df = x_dfs[i].melt(id_vars=['Country Name'],
                                value_vars=[str(i) for i in range(1960,2021)],
                                var_name='year',value_name='x')
        x_df['year'] = x_df['year'].astype(int)
        df = pd.merge(left=x_df.rename(columns={'Country Name':'country'}),
                      right=y_df,
                      on=['country','year'])
        df.dropna(axis=0,how='any',inplace=True)
        title = 'Percentage of food lost against ' + xlabels[i]
        print(title + ': ' + str(stats.pearsonr(df['x'],df['loss'])))
        sns.scatterplot(x=df['x'],y=df['loss'],data=df,alpha=0.2,ax=axs[i])
        #df.plot(kind='scatter',x='x',y='loss',alpha=0.2,ax=axs[i])
        axs[i].set_xlabel(xlabels[i])
        axs[i].set_ylabel('Percentage of food lost')
        axs[i].set_title(title)
    plt.show()
```

```python
def white_bg_color_func(word, font_size, position, orientation, random_state=None,
                    **kwargs): # some colours in the default colour scheme are hard to see on white
    return "hsl({}, {}%, {}%)".format(np.random.randint(120,270),
                                      np.random.randint(70,90),
                                      np.random.randint(20,60)) 

def eda_q4e_wordcloud(col):
    col = col.copy()
    col = col.sample(frac=1) # randomly shuffle because only the first 2000 words will be taken
    wc_text = ''
    for c in col.values:
        if c != 'Unknown':
            wc_text = wc_text + c.lower() + ' '
    wc = WordCloud(
        background_color='white',
        max_words=2000,
        color_func=white_bg_color_func,
        collocations=False
    )
    wc.generate(wc_text)
    plt.figure(figsize=(15,10))
    plt.imshow(wc, interpolation='bilinear')
    plt.axis('off')
    plt.show()
```

Q4C: tried to plot with crop group as hue, but it was quite difficult to see any patterns in the colours

```python
y_df = df_waste.groupby(['country','year','crop_group'])[['loss']].median().reset_index()
xlabels = ['agricultural area','percentage area for agriculture','aquaculture production',
          'food imports','food exports','net food trade','GDP per capita','population size']
x_dfs = [df_agri, df_agri_pc, df_aqua, df_food_in, df_food_out, net_trade, df_gdp, df_pop]

fig, axs = plt.subplots(8, 1, figsize=(15,40))

for i in range(8):
    x_df = x_dfs[i].melt(id_vars=['Country Name'],
                            value_vars=[str(i) for i in range(1960,2021)],
                            var_name='year',value_name='x')
    x_df['year'] = x_df['year'].astype(int)
    df = pd.merge(left=x_df.rename(columns={'Country Name':'country'}),
                  right=y_df,
                  on=['country','year'])
    df.dropna(axis=0,how='any',inplace=True)
    title = 'Percentage of food lost against ' + xlabels[i]
    print(title + ': ' + str(stats.pearsonr(df['x'],df['loss'])))
    sns.scatterplot(x=df['x'],y=df['loss'],data=df,alpha=0.2,ax=axs[i],hue='crop_group')
    #df.plot(kind='scatter',x='x',y='loss',alpha=0.2,ax=axs[i])
    axs[i].set_xlabel(xlabels[i])
    axs[i].set_ylabel('Percentage of food lost')
    axs[i].set_title(title)
plt.show()
```
